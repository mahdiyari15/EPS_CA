```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom, norm
```

part a

```python
n = 270
p = 0.3
x = np.arange(0, n)
y = binom.pmf(x, n, p)
plt.bar(x, y, color='blue')
plt.title("Binomial Distribution")
plt.xlabel("Number of x")
plt.ylabel("Probability Mass Function (PMF)")
plt.show()
```

part b

```python
mean = n * p
sigma = np.sqrt(n * p * (1 - p))
est = (x - mean) / sigma
plt.bar(est, y, color='blue', edgecolor='red', width=0.05)
plt.title("Standardized Binomial Distribution")
plt.xlabel("Standardized Values (z)")
plt.ylabel("Probability Mass Function (PMF)")
plt.show()
```

part c

```python
# x_normal = np.linspace(-4,4,1000)
y_normal = norm.pdf(est)
tmp = sum(y_normal)
print(tmp)
plt.figure(figsize=(12, 6))
plt.bar(est, y, color='lightcoral', edgecolor='black', width=0.05, label='Standardized Binomial')
plt.plot(est, y_normal, color='blue', linewidth=2, label='Standard Normal')
plt.title("Comparison of Standardized Binomial and Standard Normal Distribution")
plt.xlabel("Standardized Values (z)")
plt.ylabel("Probability Density / Mass")
```
