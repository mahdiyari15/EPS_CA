```python
def probability(number_of_zeros, number_of_ones, p) :
    probability_of_one = (p ** number_of_ones * (1 - p) ** number_of_zeros)
    probability_of_zero = (p ** number_of_zeros * (1 - p) ** number_of_ones)
    if number_of_ones > number_of_zeros:
        up = probability_of_one
    else : 
        up = probability_of_zero
    return (up / (probability_of_one + probability_of_zero))
scenarios = [[4, 8, 0.7], [2, 10, 0.7], [4, 8, 0.3], [3, 9, 0.5], [7, 5, 0.5]]
for scenario in scenarios:
    print(probability(scenario[0], scenario[1], scenario[2]))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import random
values = [1, 0]  
n_values = list(range(2, 51))
p_values = [i / 100 for i in range(0, 101, 10)]
heatmap_data = np.zeros((len(p_values), len(n_values)))
for i, p in enumerate(p_values):
    prob = [p, 1 - p]
    for j, n in enumerate(n_values):
        accuracy = 0
        for _ in range(1000):
            votes = random.choices(values, prob, k=n)
            ones = votes.count(1)
            zeros = votes.count(0)
            if ones > zeros:
                accuracy += 1
        heatmap_data[i, j] = accuracy / 1000
plt.figure(figsize=(12, 8))
sns.heatmap(heatmap_data, xticklabels=n_values, yticklabels=[round(p, 1) for p in p_values],
cmap='YlGnBu', annot=True, fmt=".1f", annot_kws={"size": 8}, cbar_kws={'label': 'Accuracy'},
linewidths=.5, linecolor='black')
plt.xticks(rotation=45, ha='right', fontsize=10)
plt.yticks(fontsize=10)
plt.xlabel('Number of Voters (n)', fontsize=12)
plt.ylabel('Voter Accuracy (p)', fontsize=12)
plt.title('Effect of Voter Accuracy (p) and Number of Voters (n) on Majority Vote Accuracy', fontsize=14)
plt.tight_layout()
plt.show()
```
