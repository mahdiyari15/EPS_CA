ANSWER 1

Ignoring means that when we encounter a word that has not been seen in the previous training set (BoW), we remove that word from the computation. That is, we do not calculate the probability of that word at all in the model.

Consequences of ignoring words:
Reduced prediction accuracy: Ignoring unknown words may result in reduced model prediction accuracy, as any word in an email can provide useful information to determine whether an email is spam or not. When unknown words are ignored, some information is lost.
Low effect in limited cases: If the number of unknown words in the email is small, ignoring them will not have much effect on the prediction. But if a large number of email words are not in the BoW, the model will not have nearly enough information to make accurate predictions.
More stability: This method may be more stable than the zero-likelihood method, because the total probability of the model does not become zero due to the presence of unknown words.

If there is a word in the email that does not appear in the BoW, we consider the probability of that word to be zero. This means we assume the word has never been seen in the spam or non-spam category.
Zero consequences of considering probability:
Zeroing the total probability: In the simple Bayes model, the probabilities of the words in the email are multiplied together. If the probability of a word occurring is zero (a word that is not in the BoW), the total predicted probability for that email is zero, since any number multiplied by zero is zero. This causes the model to fail to make any valid predictions for that email, even if most of the words in the email correctly indicate that the email is spam or non-spam.
Incorrect predictions: Zeroing the probability of even a single word will cause the model to make incorrect predictions. Even if all other words indicate that the email is spam, the presence of an unknown word makes the total probability zero and finally the model may wrongly classify the email as non-spam or spam.

Using Laplace Smoothing helps us to avoid the problem of zero probabilities. Instead of considering the probability of unknown words to be zero, by adding one unit to the number of occurrences of each word (even unknown words), the probability of occurrence of those words will have at least a non-zero value (however small).

```python
import pandas as pd
import numpy as np
import re
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer

def preprocess_text(text):
    text = text.lower()
    text = re.sub(r'[^\w\s]', '', text)
    text = re.sub(r'\d+', '', text)
    return text

def predict_email_class(email, spam_word_prob, non_spam_word_prob, P_spam, P_non_spam, word_to_index):
    words = email.split()
    indices = []
    for word in words:
        if word in word_to_index:
            indices.append(word_to_index[word])
    P_spam *= np.prod(spam_word_prob[indices])
    P_non_spam *= np.prod(non_spam_word_prob[indices])
    return 1 if P_spam > P_non_spam else 0

def accuracy_score(y_true, y_pred):
    total_samples = len(y_true)
    correct_predictions = 0
    for i in range(total_samples):
        if y_true.iloc[i] == y_pred[i]:
            correct_predictions += 1
    accuracy = correct_predictions / total_samples
    return accuracy

file = pd.read_csv('emails.csv')
if 'text' in file.columns:
    file['text'] = file['text'].apply(preprocess_text)
train1, test1, train2, test2 = train_test_split(
    file['text'], file['spam'], test_size=0.3, random_state=532
)
vectorizer = CountVectorizer()
train1_bow = vectorizer.fit_transform(train1)
test1_bow = vectorizer.transform(test1)
features = vectorizer.get_feature_names_out()
number_of_features = len(features)
number_of_spam = sum(train2)
number_of_non_spam = len(train2) - number_of_spam
spam_counts = np.zeros(number_of_features)
non_spam_counts = np.zeros(number_of_features)
for i in range(len(train2)):
    if train2.iloc[i] == 1:
        spam_counts += train1_bow[i].toarray()[0]
    else:
        non_spam_counts += train1_bow[i].toarray()[0]
spam_word_prob = (spam_counts + 1) / (np.sum(spam_counts) + number_of_features)
non_spam_word_prob = (non_spam_counts + 1) / (np.sum(non_spam_counts) + number_of_features)
P_spam = number_of_spam / len(train2)
P_non_spam = number_of_non_spam / len(train2)
word_to_index = {}
for i, word in enumerate(features):
    word_to_index[word] = i
predictions = []
for email in test1:
    prediction = predict_email_class(email, spam_word_prob, non_spam_word_prob, P_spam, P_non_spam, word_to_index)
    predictions.append(prediction)
accuracy = accuracy_score(test2, predictions)
print(f"Accuracy of the model: {accuracy:.8f}")
```

ANSWER

In the Bayes classification algorithm, when the text of the message is long and the words are likely to be mixed together, a main problem arises, which we call the zero number problem or the problem of small number (underflow). This is because by multiplying small probabilities, the result can quickly approach very small values ​​that may eventually approach zero. Consequently, this can negatively affect the accuracy of model predictions.
Report the results of using the logarithm:
Higher accuracy: By using the logarithm, the accuracy of the model increases significantly.
Stability: Compared to long texts, the model will be numerically more stable and its accuracy will be less affected by small probabilities.
These changes can help improve the quality of predictions and provide more efficiency, especially in scenarios with long texts and rare words.
