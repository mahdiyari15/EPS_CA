## Calculate Correlation
complete `correlation` function that calculate the correlation matrix of a data fram.

**Note:** you are not allowed to use `pandas.core.frame.DataFrame.corr()` method

```python
def correlation(data: pd.DataFrame) -> pd.DataFrame:
    """
    Calculates the correlation matrix from a given dataset.

    Args:
        data (pandas.core.frame.DataFrame): The dataset, where each row represents an observation and each column represents a feature.

    Returns:
        correlation_matrix (pandas.core.frame.DataFrame): The correlation matrix.
    """
    ######################  TODO  ########################
    n = len(data.columns)
    matrix = np.zeros((n, n))
    for i in range(n):
        for j in range(n):
            x = data.iloc[:, i]
            y = data.iloc[:, j]
            E_x = np.mean(x)
            E_y = np.mean(y)
            E_xy = np.mean(x * y)
            cov_xy = E_xy - E_x * E_y
            var_x = np.var(x)
            var_y = np.var(y)
            correlation_of_xy = cov_xy / np.sqrt(var_x * var_y)
            matrix[i, j] = correlation_of_xy
    correlation_matrix = pd.DataFrame(matrix, columns=data.columns, index=data.columns)
    return correlation_matrix
data = pd.DataFrame({
    'x': [45, 37, 42, 35, 39],
    'y': [38, 31, 26, 28, 33],
    'z': [10, 15, 17, 21, 12]
})
correlation_matrix = correlation(data)
print(correlation_matrix)
    ######################  TODO  ########################
```
